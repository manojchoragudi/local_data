```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
df_bulk = spark.read \
    .option("mode", "PERMISSIVE") \
    .json("/Volumes/local_cataog/default/local_volume/bulk_rides.json")

df_bulk.printSchema()
```

```python
df_bulk = spark.read \
    .option("multiline", "true") \
    .option("inferSchema", "true") \
    .json("/Volumes/local_cataog/default/local_volume/bulk_rides.json")

display(df_bulk)
```

**Standardize the column names**

```python
df_bulk = df_bulk.toDF(*[c.strip().lower() for c in df_bulk.columns])
```

**Trim whitespace in all string columns**

```python
from pyspark.sql import functions as F
for c, t in df_bulk.dtypes:
    if t == "string":
        df_bulk = df_bulk.withColumn(c, F.trim(F.col(c)))
```

**Drop rows where the key is null**

```python
df_bulk = df_bulk.filter(F.col("ride_id").isNotNull())
```

**Remove duplicate rows**

```python
df_bulk = df_bulk.dropDuplicates(["ride_id"])
```
